```python
data = [
    (1, "Chandana", 50000),
    (2, "Rahul",    60000),
    (3, "Priya",    55000),
]
columns = ["id", "name", "salary"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
df.printSchema()   # column names, types, nullability
print(df.columns)  # ['id', 'name', 'salary']  (a Python list, no job runs)
print(df.count())  # 3  (an action: runs a Spark job)
df.display()       # Databricks' rich, sortable table with charts
```
