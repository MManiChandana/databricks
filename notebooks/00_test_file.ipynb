```python
d = [
    (1, "Chandana", 500),
    (2, "Rahul", 750),
    (3, "Priya", 1200)
]
c = ["customer_id", "customer_name", "total_spent"]
df = spark.createDataFrame(data=d, schema=c)
df.write.mode("overwrite").saveAsTable("customer")
```
